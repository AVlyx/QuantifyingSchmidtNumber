```python
from toqito.state_props import is_ppt
from toqito.rand import random_density_matrix
from tqdm import tqdm
from math import prod

dims = (2, 4)


def random_npt():
    while True:
        m = random_density_matrix(prod(dims))
        if not is_ppt(m, dim=list(dims)):
            return m


def random_ppt():
    while True:
        m = random_density_matrix(prod(dims))
        if is_ppt(m, dim=list(dims)):
            return m


def random_close_to_edge_npt():
    rhonpt = random_npt()
    rhoppt = random_ppt()
    for _ in range(10):
        curr = (rhonpt + rhoppt) / 2
        if is_ppt(curr, dim=list(dims)):
            rhoppt = curr
        else:
            rhonpt = curr
    return rhonpt
```

```python
from SDP_utils.sdp import SDP

TOL = 0

detected_at_X_copies = {"Not detected": 0, 3: 0, 4: 0, 5: 0}


def hierarchy(rho, kmax: int):
    for i in range(3, kmax + 1):
        obj, _ = SDP(2, 3, rho, dims)
        objectives[i] = obj
        if obj >= TOL:
            detected_at_X_copies[i] += 1
            print(objectives)
            print(detected_at_X_copies)
            return
    print(objectives)
    print(detected_at_X_copies)    


for i in tqdm(range(1000)):
    rho = random_close_to_edge_npt()
    obj, _ = SDP(2, 2, rho, dims)
    if obj >= TOL:
        continue
    detected_at_X_copies["Not detected"] += 1
    objectives = {2: obj}

    hierarchy(rho, 3)

print(detected_at_X_copies)
```
