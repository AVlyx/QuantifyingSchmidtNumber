```python
k = 3
from math import comb
def rootsTest(x, r, t, k, P_lambda):
    s = 0
    for i in range(k+1):
        s += comb(t, i) * comb(r-t, k-i) * (x/t)**i * ((1-x)/(r-t))**(k-i) 
    return s - P_lambda
```
