```python
from SDP_utils.combinatorics import dim_sym_kd


sb2 = """ \\begin{tabular}{c|c c c c}
        $k \\setminus d$ & $\\C^2\\times\\C^2$ & $\\C^3\\times\\C^3$ & $\\C^4\\times\\C^4$ & $\\C^5\\times\\C^5$ \\\\ \n"""
for k in range(2, 6):
    sb2 += f"       ${k}$ & "
    for i in range(2,6):
        d= i*i
        # val = dim_sym_kd(k, d)
        # val = d**k
        val = dim_sym_kd(k//2, d) * dim_sym_kd(k-k//2,d)
        if val < 300:
            sb2 += f"${val}$ {"&" if i<5 else ""} "
        else:
            sb2 += f"${val:.0e}$ {"&" if i<5 else ""} "
    sb2 += "\\\\\n"
sb2 += "\\end{tabular}"

print(sb2)
```
