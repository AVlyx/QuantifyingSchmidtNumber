```python
from pathlib import Path
from notebook_utils.load_store_results import load_all_results, SdpResult
from SDP_utils.E_t import Et_lower

folder = "test"

for file in Path(f"./sdp_results/{folder}").rglob("*.jsonl"):
    file = file.name
    results = load_all_results(folder, file)
    for res in results:
        if res.minSchmidtNumber == 2:
            res.Et_lower = Et_lower([1,1], res.objective, 2)
        res.dump_to_jsonl(f"sdp_results/{folder.lower()}2/{file}")
```
