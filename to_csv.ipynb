```python
histogram_files: list[tuple[str, str, str]] = [
    ("noise", "Chessboard_with_noise.jsonl", "Chessboard"),
    ("noise", "Cross_hatch_with_noise.jsonl", "Cross Hatch"),
    ("noise", "Gen tiles1_with_noise.jsonl", "Gen Tiles 1"),
    ("noise", "Gen tiles2_with_noise.jsonl", "Gen Tiles 1"),
    ("noise", "Ncomms6297_with_noise.jsonl", "Ncomms6297"),
    ("noise", "Piani_with_noise.jsonl", "Piani"),
    ("noise", "Pyramid_with_noise.jsonl", "Pyramid"),
    ("noise", "Steering_with_noise.jsonl", "Steering"),
    ("noise", "Tiles_with_noise.jsonl", "Tiles"),
    ("noise", "Yu_Oh_4x4_at_0.33_0.33_with_noise.jsonl", "Yu-Oh"),
    ("Parametric", "Breur.jsonl", "Breuer"),
    ("Parametric", "c3c3_Horodecki.jsonl", "Horodecki"),
]
```

```python
from notebook_utils.load_store_results import load_all_results

histogram_data: list[tuple[str, float]] = []

for folder, file, name in histogram_files:
    data = load_all_results(folder, file)
    max_E_t = max(data, key=lambda x: x.Et_lower[0])
    histogram_data.append((name, max_E_t.Et_lower[0]))

import csv

with open("./csv_for_plots/histogram.csv", "w", newline="") as f:
    writer = csv.writer(f, delimiter=";")
    writer.writerow(["state", "E1_lower"])
    writer.writerows(histogram_data)
```

```python
line_plot_files: list[tuple[str, str, str]] = [
    ("convex", "convex_Chessboard_Cross_hatch.jsonl", "Chessboard - Cross Hatch"),
    ("convex", "convex_Chessboard_Pyramid.jsonl", "Chessboard - Pyramid"),
    ("convex", "convex_Cross_hatch_Tiles.jsonl", "Cross Hatch - Tiles"),
    ("convex", "convex_Pyramid_Steering.jsonl", "Pyramid - Steering"),
]
```

```python
from notebook_utils.load_store_results import load_results_in_range
import numpy as np

lineplot_data = []
range_ = (0, 1, 0.02)
for folder, file, name in line_plot_files:
    data = load_results_in_range(folder, file, range_)
    print(data[-1])
    assert len(data) == (range_[1] - range_[0]) / range_[2] + 1  # Would otherwise cause issues in the latex
    lineplot_data.append(
        (
            name,
            [(point.Et_lower[0] if point.Et_lower[0] else 1e-8, point.separability) for point in data],
        )
    )

p = list(np.arange(range_[0], range_[1] + range_[2], range_[2]))


for name, points in lineplot_data:
    with open(f"./csv_for_plots/linePlot/{name}.csv", "w", newline="") as f:
        writer = csv.writer(f, delimiter=";")
        writer.writerow(["p", "E1_lower", "separability"])
        for p_val, (e1_lower, sep) in zip(p, points):
            writer.writerow([p_val, e1_lower, sep])
```
