```python
from notebook_utils.load_store_results import load_results_in_range
import numpy as np

lineplot_data = []
range_ = (0, 1, 0.04)
for folder, file, name in line_plot_files:
    data = load_results_in_range(folder, file, (0, 1, 0.04))
    print(data[-1])
    assert len(data) == (range_[1] - range_[0]) / range_[2] + 1  # Would otherwise cause issues in the latex
    lineplot_data.append(
        (
            name,
            [(point.Et_lower[0] if point.Et_lower[0] else 1e-8, point.separability) for point in data],
        )
    )

p = list(np.arange(range_[0], range_[1] + range_[2], range_[2]))


for name, points in lineplot_data:
    with open(f"./csv_for_plots/linePlot/{name}.csv", "w", newline="") as f:
        writer = csv.writer(f, delimiter=";")
        writer.writerow(["p", "E1_lower", "separability"])
        for p_val, (e1_lower, sep) in zip(p, points):
            writer.writerow([p_val, e1_lower, sep])
```
